# Experiment 16: Prompt Evaluation & Ethics Workshop

Regenerated as a standalone notebook using Google Gemini API.

```python

import os, json, re, getpass
try:
    from google import genai
except ImportError as exc:
    raise ImportError("Install the Gemini SDK first: %pip install -q google-genai") from exc

# Use a new Gemini API key. Input is hidden and not saved in this notebook.
API_KEY = os.getenv("GEMINI_API_KEY") or getpass.getpass("Enter your Gemini API key (input hidden): ")
MODEL = "gemini-2.5-flash"
client = genai.Client(api_key=API_KEY)

def ask_llm(prompt, system="You are a helpful assistant.", temperature=0.2):
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config={"system_instruction": system, "temperature": temperature}
    )
    return response.text or ""

print("Gemini client configured. Model:", MODEL)
```

## Experiment Program

```python
samples = [
 {"question":"What causes sea level rise?", "context":"Melting land ice and warming seawater raise sea levels.", "reference":"Melting land ice and warming seawater raise sea levels."},
 {"question":"How do renewables help?", "context":"Solar and wind can reduce emissions when replacing fossil fuels.", "reference":"Renewables can reduce emissions when replacing fossil fuels."}
]
for item in samples:
    prompt = ("Answer only from the context. Do not speculate. If unsupported, say I don't know.\\n"
              f"Context: {item['context']}\\nQuestion: {item['question']}")
    answer = ask_llm(prompt)
    print("\\nQuestion:", item["question"], "\\nAnswer:", answer)
print("\\nEthics checklist: prompt injection, hallucinations, privacy, bias, transparent limitations, human review.")
```
